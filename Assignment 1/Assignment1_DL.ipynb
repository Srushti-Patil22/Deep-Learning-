```python
# 1. Import Libraries

import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np
```

```python
# 2. Load MNIST Dataset and Split into Train and Test Sets

(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.mnist.load_data()

print("Training Images Shape :", train_images.shape)
print("Training Labels Shape :", train_labels.shape)
print("Testing Images Shape  :", test_images.shape)
print("Testing Labels Shape  :", test_labels.shape)
```

```python
# 3. Data Preprocessing (Normalization)

train_images = train_images / 255.0
test_images = test_images / 255.0

print("Minimum Pixel Value :", train_images.min())
print("Maximum Pixel Value :", train_images.max())
```

```python
# 4. Train-Test Split Information

print("Training Samples :", len(train_images))
print("Testing Samples  :", len(test_images))
```

```python
# 5. Visualize Dataset Information

print("Training Image Shape :", train_images.shape)
print("Testing Image Shape  :", test_images.shape)

print("\nFirst 10 Labels:")
print(train_labels[:10])
```

```python
# 6. Display First Image

plt.imshow(train_images[0], cmap='gray')
plt.title("Label : {}".format(train_labels[0]))
plt.axis('off')
plt.show()
```

```python
# 7. Display First 10 Images

plt.figure(figsize=(10,5))

for i in range(10):
    plt.subplot(2,5,i+1)
    plt.imshow(train_images[i], cmap='gray')
    plt.title(train_labels[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# 8. Defining Neural Network Model

my_model = tf.keras.models.Sequential()

# Input Layer
my_model.add(tf.keras.layers.Flatten(input_shape=(28, 28)))

# Hidden Layer
my_model.add(tf.keras.layers.Dense(128, activation='relu'))

# Output Layer
my_model.add(tf.keras.layers.Dense(10, activation='softmax'))
```

```python
# 7. Compiling the Model

my_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
# 8. Training the Model

history = my_model.fit(
    train_images,
    train_labels,
    epochs=3
)
```
